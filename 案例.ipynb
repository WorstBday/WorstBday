```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

# 数据
time_labels = [
    "2010年3月", "2010年6月", "2010年9月", "2010年12月",
    "2011年3月", "2011年6月", "2011年9月", "2011年12月",
    "2012年3月", "2012年6月", "2012年9月", "2012年12月",
    "2013年前2月"
]
# 新开工项目同比增速%（蓝色折线）
rate_new_project = [11, -7, -6, -7, -10, -4, 0, -2, 16, -1, 14, 2, 11]
# 新开工项目计划投资同比增速%（红色折线）
rate_invest = [35, 23, 26, 26, -4, 32, 28, 25, 23, 24, 29, 29, 11]

x = np.arange(len(time_labels))

fig, ax = plt.subplots(figsize=(14,7), dpi=120)

# 绘制两条折线
ax.plot(x, rate_new_project, color="#4472c4", linewidth=3, label="新开工项目同比增速%")
ax.plot(x, rate_invest, color="#c0392b", linewidth=3, label="新开工项目计划投资同比增速%")

# 坐标轴设置
ax.set_ylim(-15, 40)
ax.set_yticks([-15,-10,-5,0,5,10,15,20,25,30,35,40])
ax.set_xticks(x)
ax.set_xticklabels(time_labels, rotation=80)
ax.legend(loc="upper center", ncol=2, frameon=False)
ax.grid(axis="y", alpha=0.4)

plt.title("图表 6 12年计划投资项目预计将在13年才会实质动工", fontsize=15, pad=20, color="#c02020")
plt.tight_layout()
plt.savefig("invest_project_trend.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

# 数据
time_labels = [
    "2011年1月", "2011年3月", "2011年5月", "2011年7月", "2011年9月", "2011年11月",
    "2012年1月", "2012年3月", "2012年5月", "2012年7月", "2012年9月", "2012年11月",
    "2013年1月"
]
# 铁路投资同比增速（红色折线，左坐标轴）
rail_growth = [55, 30, -10, -25, -70, -80, -85, -40, -20, 0, 130, 40, 60]
# 公路投资同比增速（蓝色折线，右坐标轴）
road_growth = [10, 30, 10, 12, 12, -5, -55, 30, -10, 0, 10, 10, 45]

x = np.arange(len(time_labels))

fig, ax1 = plt.subplots(figsize=(14,7), dpi=120)

# 红色折线：铁路投资同比增速，左Y轴
ax1.plot(x, rail_growth, color="#c0392b", linewidth=3, label="铁路投资同比增速")
ax1.set_ylabel("铁路投资同比增速", fontsize=12)
ax1.set_ylim(-100, 300)
ax1.set_yticks([-100,-50,0,50,100,150,200,250,300])
ax1.set_yticklabels(["-100%","-50%","0%","50%","100%","150%","200%","250%","300%"])
ax1.set_xticks(x)
ax1.set_xticklabels(time_labels, rotation=80)

# 蓝色折线：公路投资同比增速，右Y轴
ax2 = ax1.twinx()
ax2.plot(x, road_growth, color="#4472c4", linewidth=3, label="公路投资同比增速")
ax2.set_ylabel("公路投资同比增速", fontsize=12)
ax2.set_ylim(-60, 60)
ax2.set_yticks([-60,-40,-20,0,20,40,60])
ax2.set_yticklabels(["-60%","-40%","-20%","0%","20%","40%","60%"])

# 合并图例
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc="upper center", ncol=2, frameon=False)

plt.title("图表 7 铁路以及公路投资见底回升趋势逐步确立", fontsize=15, pad=20, color="#c02020")
plt.tight_layout()
plt.savefig("rail_road_invest.png", dpi=150, bbox_inches="tight")
plt.show()
```
