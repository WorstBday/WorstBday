```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.colors as mcolors
import platform

system_name = platform.system()
if system_name == "Windows":
    plt.rcParams['font.sans-serif'] = ['SimHei']
elif system_name == "Darwin":
    plt.rcParams['font.sans-serif'] = ['Arial Unicode MS']
else:
    plt.rcParams['font.sans-serif'] = ['WenQuanYi Micro Hei']
plt.rcParams['axes.unicode_minus'] = False

fig = plt.figure(figsize=(12, 6))
bg_color = '#121633'
fig.patch.set_facecolor(bg_color)

ax = fig.add_axes([0.3, 0.05, 0.5, 0.75])
ax.set_facecolor(bg_color)
ax.set_xlim(-1.2, 1.2)
ax.set_ylim(-1.2, 1.2)
ax.axis('off')

completed = 85
remaining = 100 - completed

radius = 1.0
width = 0.12
gap_angle = 90
remaining_angle = (remaining / 100) * 360
completed_angle = (completed / 100) * 360

theta = np.linspace(0, remaining_angle, 100)
x_outer = (radius + width/2) * np.cos(np.radians(theta + gap_angle))
y_outer = (radius + width/2) * np.sin(np.radians(theta + gap_angle))
x_inner = (radius - width/2) * np.cos(np.radians(theta[::-1] + gap_angle))
y_inner = (radius - width/2) * np.sin(np.radians(theta[::-1] + gap_angle))
ax.fill(np.concatenate([x_outer, x_inner]), np.concatenate([y_outer, y_inner]), color='#1c253b', zorder=2)

cmap = mcolors.LinearSegmentedColormap.from_list("custom_gradient", ["#8b3f9e", "#e8395a"])
N = 100
for i in range(N):
    angle_start = gap_angle + remaining_angle + (i / N) * completed_angle
    angle_end = gap_angle + remaining_angle + ((i + 1) / N) * completed_angle
    color = cmap(i / N)
    
    theta_seg = np.linspace(angle_start, angle_end, 10)
    x_outer_seg = (radius + width/2) * np.cos(np.radians(theta_seg))
    y_outer_seg = (radius + width/2) * np.sin(np.radians(theta_seg))
    x_inner_seg = (radius - width/2) * np.cos(np.radians(theta_seg[::-1]))
    y_inner_seg = (radius - width/2) * np.sin(np.radians(theta_seg[::-1]))
    
    ax.fill(np.concatenate([x_outer_seg, x_inner_seg]), 
            np.concatenate([y_outer_seg, y_inner_seg]), 
            color=color, zorder=3, linewidth=0)

ax.text(0, 0.1, f'{completed}%', ha='center', va='center', color='#ffffff', fontsize=45, zorder=5)
ax.text(0, -0.2, '目标完成率', ha='center', va='center', color='#d0d0d0', fontsize=16, zorder=5)

fig.text(0.12, 0.88, '2022年上半年目标完成率', fontsize=26, fontweight='bold', color='#ffffff', ha='left')
fig.text(0.12, 0.80, '截至6月30日销售目标总体完成率达到85%', fontsize=15, color='#d0d0d0', ha='left')
fig.text(0.12, 0.05, '*注：数据来源于公司销售系统', fontsize=10, color='#888888', ha='left')

plt.show()
```
